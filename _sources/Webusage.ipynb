##**Filter Data Berdasarkan Ekstensi .html**

```python
import pandas as pd
from IPython.display import display

file_name = "webusage.csv" # Mengoreksi nama file dari "webuage.csv" menjadi "webusage.csv"

print(f"Memuat data dari {file_name}...")

# Gunakan pemisah koma karena header menunjukkan koma
df = pd.read_csv(file_name, sep=',')

print("\nDaftar kolom dalam dataset:")
print(df.columns.tolist())

print("\nInfo data asli:")
df.info()

print("\nData asli (head):")
display(df.head())

# Tentukan kolom target yang sekarang pasti terbaca
target_column = "Request URI"

if target_column not in df.columns:
    print(f"\n❌ Kolom '{target_column}' tidak ditemukan!")
else:
    print(f"\nMemfilter data pada kolom '{target_column}' yang berakhiran '.html'...")

    filtered_df = df[df[target_column].str.endswith('.html', na=False)].copy()

    output_filename = "html_requests_only.csv"
    filtered_df.to_csv(output_filename, index=False)

    print(f"\n✅ Berhasil menyimpan hasil filter ke '{output_filename}'.")
    display(filtered_df.head())
```

## **Filter Dengan Status 200**

```python
file_name = "html_requests_only.csv"

# 1. Muat dataset
print(f"Memuat data dari {file_name}...")
df = pd.read_csv(file_name)

# 2. Tentukan kolom target
target_column = "Status"

# 3. Periksa apakah kolom target ada
if target_column not in df.columns:
    print(f"Error: Kolom '{target_column}' tidak ditemukan di CSV.")
else:
    # 4. Periksa apakah kolomnya numerik
    if pd.api.types.is_numeric_dtype(df[target_column]):

        # 5. Filter baris di mana 'Status' = 200
        print(f"\nMemfilter data untuk baris di mana '{target_column}' adalah 200...")
        filtered_df = df[df[target_column] == 200].copy()

        # 6. Simpan data hasil filter
        output_filename = "status_200_only.csv"
        filtered_df.to_csv(output_filename, index=False)

        # 7. Tampilkan hasil dalam kotak tabel
        print(f"\nBerhasil memfilter data dan menyimpannya ke '{output_filename}'.")
        print("Pratinjau data yang telah difilter:")
        display(filtered_df.head())

    else:
        print(f"Error: Kolom '{target_column}' bukan numerik. Tidak dapat memfilter berdasarkan nilai 200.")
```

## **Membersihkan Format Waktu di Kolom Request Time menjadi tanggal dan waktunya**

```python
file_name = "method_get_only.csv"
target_column = "Request time"

# 1. Muat dataset
print(f"Memuat data dari {file_name}...")
df = pd.read_csv(file_name)

# 2. Periksa apakah kolom target ada
if target_column not in df.columns:
    print(f"Error: Kolom '{target_column}' tidak ditemukan di CSV.")
else:
    # 3. Tampilkan contoh format asli (opsional)
    print(f"Format asli (baris pertama): {df[target_column].iloc[0]}")

    # 4. Bersihkan format datetime
    print(f"Mengubah format kolom '{target_column}'...")
    df[target_column] = df[target_column].str.replace('T', ' ', regex=False)
    df[target_column] = df[target_column].str.replace('Z', '', regex=False)

    # 5. Simpan ke CSV baru
    output_filename = "request_time_cleaned.csv"
    df.to_csv(output_filename, index=False)

    # 6. Tampilkan hasil dalam tabel rapi
    print(f"\nBerhasil mengubah format data dan menyimpannya ke '{output_filename}'.")
    print("Pratinjau data yang telah diubah:")
    display(df.head())
```

## **Membuat Kolom pada session berdasarkan Host, Mengurutkan Berdasarkan Waktu, dan Menyimpan Tabel Final**

```python
file_name = "time_converted_NASA.csv"
col_ip = "host"
col_time = "datetime_readable"
new_col_name = "SESSION"

print(f"Memuat data dari {file_name}...")

# 1. Baca file dengan PEMISAH OTOMATIS (jangan pakai ; )
df = pd.read_csv(file_name)

print("Kolom terdeteksi:", df.columns.tolist())

# 2. Cek kolom wajib
if col_ip not in df.columns or col_time not in df.columns:
    raise ValueError(f"Kolom '{col_ip}' atau '{col_time}' tidak ditemukan!")

# 3. Konversi kolom waktu ke datetime
df[col_time] = pd.to_datetime(df[col_time], errors='coerce')

# 4. Buat kolom SESSION
df[new_col_name] = df[col_ip].astype('category').cat.codes + 1

# 5. Sortir berdasarkan SESSION & waktu
df_sorted = df.sort_values(by=[new_col_name, col_time])

# 6. Hapus kolom time lama jika masih ada
if "time" in df_sorted.columns:
    df_sorted = df_sorted.drop(columns=["time"])

# 7. Simpan
output_filename = "session_data_full_NASA.csv"
df_sorted.to_csv(output_filename, index=False)

print(f"\nSukses! File disimpan sebagai {output_filename}")
display(df_sorted[[new_col_name, col_ip, col_time, "url"]].head())
```

## **Konversi URL Menjadi Label Huruf**

```python
def num_to_letters(n):
    """Konversi angka ke huruf: 0→A, 1→B, ..., 25→Z, 26→AA, dst."""
    result = ""
    n += 1
    while n > 0:
        n -= 1
        result = chr(65 + (n % 26)) + result
        n //= 26
    return result

# --- Hasil Langsung ---
[num_to_letters(i) for i in range(10)]
```

## **Pembentukan Web Usage Session**

```python
def num_to_letters(n):
    string = ""
    while n >= 0:
        string = chr(n % 26 + ord('A')) + string
        n = n // 26 - 1
    return string

def process_web_usage(df):
    """
    Membentuk sesi web usage 20 menit per host.
    URL diganti dengan huruf A-Z.
    """

    # Buat mapping URL → huruf
    unique_urls = df["url"].unique() # Corrected 'URL' to 'url'
    url_map = {url: num_to_letters(i) for i, url in enumerate(unique_urls)}
    df["URL_Label"] = df["url"].map(url_map) # Corrected 'URL' to 'url'

    df = df.sort_values(by=["Host", "Timestamp"])

    results = []

    for host, group in df.groupby("Host"):
        group = group.sort_values("Timestamp").reset_index(drop=True)

        sessions = []
        cur_session = []
        prev_time = None

        for _, row in group.iterrows():
            ts = row["Timestamp"]
            label = row["URL_Label"]

            if prev_time is None:
                cur_session.append((ts, label))
            else:
                delta = (ts - prev_time).total_seconds() / 60

                # aturan sesi 20 menit
                if delta > 20:
                    sessions.append(cur_session)
                    cur_session = [(ts, label)]
                else:
                    cur_session.append((ts, label))

            prev_time = ts

        if cur_session:
            sessions.append(cur_session)

        results.append({
            "Host": host,
            "Sessions": sessions
        })

    return results, url_map


# --- Hasil Langsung ---
results, url_map = process_web_usage(df)
results[:1]   # tampilkan contoh 1 host pertama
```

## **Membentuk Tabel Web Usage (Host – Session – Timestamp – Page)**

```python
url_mapping_df = pd.DataFrame([
    {"URL": url, "Label": label}
    for url, label in url_map.items()
])

url_mapping_df
```

## **Ringkasan Total Sesi Per Host**

```python
table_rows = []

for item in results:
    host = item["Host"]
    sessions = item["Sessions"]

    for session_idx, session in enumerate(sessions, start=1):
        for ts, label in session:
            table_rows.append({
                "Host": host,
                "Session": session_idx,
                "Timestamp": ts,
                "Page": label
            })

usage_table = pd.DataFrame(table_rows)

usage_table
```

```python
session_summary = usage_table.groupby("Host")["Session"].nunique().reset_index()
session_summary.rename(columns={"Session": "Total Sessions"}, inplace=True)

session_summary
```

## **MEMBANGUN TABEL BINARY PER HALAMAN**

```python
# 5. MEMBANGUN TABEL BINARY PER HALAMAN (A=1, lainnya=0)

import pandas as pd # Import pandas for DataFrame operations

# --- Start code to ensure 'df' is available ---
# Path dataset
file_path = "data.csv"

# Load df and preprocess (copied from cell 0Z9oSXRTOQHd)
df = pd.read_csv(file_path, sep=',')

# Convert time
df["Timestamp"] = pd.to_datetime(df["time"], unit="s")

# Rename
df.rename(columns={"host": "Host"}, inplace=True)
# --- End code to ensure 'df' is available ---

def num_to_letters(n):
    string = ""
    while n >= 0:
        string = chr(n % 26 + ord('A')) + string
        n = n // 26 - 1
    return string

def process_web_usage(df):
    """
    Membentuk sesi web usage 20 menit per host.
    URL diganti dengan huruf A-Z.
    """

    # Buat mapping URL → huruf
    unique_urls = df["url"].unique()
    url_map = {url: num_to_letters(i) for i, url in enumerate(unique_urls)}
    df["URL_Label"] = df["url"].map(url_map)

    df = df.sort_values(by=["Host", "Timestamp"])

    results = []

    for host, group in df.groupby("Host"):
        group = group.sort_values("Timestamp").reset_index(drop=True)

        sessions = []
        cur_session = []
        prev_time = None

        for _, row in group.iterrows():
            ts = row["Timestamp"]
            label = row["URL_Label"]

            if prev_time is None:
                cur_session.append((ts, label))
            else:
                delta = (ts - prev_time).total_seconds() / 60

                # aturan sesi 20 menit
                if delta > 20:
                    sessions.append(cur_session)
                    cur_session = [(ts, label)]
                else:
                    cur_session.append((ts, label))

            prev_time = ts

        if cur_session:
            sessions.append(cur_session)

        results.append({
            "Host": host,
            "Sessions": sessions
        })

    return results, url_map

def build_binary_table(results, url_map):
    all_labels = sorted(url_map.values())  # A, B, C, ...
    rows = []

    for host_data in results:
        host = host_data["Host"]

        for session in host_data["Sessions"]:
            for ts, label in session:

                # buat baris kosong (semua 0)
                row = {l: 0 for l in all_labels}

                # halaman yang dikunjungi = 1
                row[label] = 1

                # tambahkan kolom lain
                row["Time"] = ts.strftime("%H:%M")
                row["Host"] = host

                rows.append(row)

    return pd.DataFrame(rows)

# Ensure 'results' and 'url_map' are defined. If not, re-generate from 'df'.
# Note: 'df' is now guaranteed to be available from the code above.
if 'results' not in locals() or 'url_map' not in locals():
    print("Warning: 'results' or 'url_map' not found. Attempting to re-generate from 'df'...")
    # 'process_web_usage' and 'num_to_letters' are now defined in this cell
    results, url_map = process_web_usage(df)

binary_table = build_binary_table(results, url_map)

print("=== Tabel Binary Halaman Dikunjungi ===")
display(binary_table.head(20))

# Simpan CSV bila diperlukan
binary_table.to_csv("binary_web_usage.csv", index=False)
```
